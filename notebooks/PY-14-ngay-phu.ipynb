Mở notebook và tìm SKU sắp đứt hàng. Ngày phủ = tồn hiện tại chia cho sức bán mỗi ngày, sức bán mỗi ngày = bán 28 ngày / 28. In một số nguyên: có bao nhiêu SKU có ngày phủ dưới 7. Rồi viết những SKU đó là gì và một kiểm tra dữ liệu trước khi chốt.

Chạy được trên Google Colab. Pandas đã có sẵn.


Bảng tồn và bán 28 ngày của Nam Việt. Ngày phủ = ton / (ban_28 / 28). SKU dưới 7 ngày phủ là nhóm cần xem trước.

A01 tồn 140, bán 280. A02 tồn 20, bán 140. B01 tồn 300, bán 140. B02 tồn 50, bán 280. C01 tồn 10, bán 70.


```python
import pandas as pd
from io import StringIO
csv = '''sku,ton,ban_28
A01,140,280
A02,20,140
B01,300,140
B02,50,280
C01,10,70'''
df = pd.read_csv(StringIO(csv))
print(df.to_string(index=False))
```

```python
# In đúng một số nguyên: số SKU có ngày phủ < 7.
```
