Mở notebook, tính fill rate từng kho của Nam Việt rồi in một số: fill rate thấp nhất, đơn vị %, làm tròn 1 chữ số thập phân. Sau đó viết kho nào cần xem trước và vì sao không được lấy trung bình của ba tỷ lệ.

Chạy được trên Google Colab. Pandas đã có sẵn.


Mỗi dòng là một đơn. Fill rate của một kho = tổng giao / tổng đặt của kho đó, nhân 100.

Bac có D01 (đặt 100, giao 92) và D02 (50, 45). Trung có D03 (80, 72) và D04 (40, 36). Nam có D05 (200, 140) và D06 (100, 80).


```python
import pandas as pd
from io import StringIO
csv = '''ma_don,kho,dat,giao
D01,Bac,100,92
D02,Bac,50,45
D03,Trung,80,72
D04,Trung,40,36
D05,Nam,200,140
D06,Nam,100,80'''
df = pd.read_csv(StringIO(csv))
print(df.to_string(index=False))
```

```python
# In đúng một số: fill rate thấp nhất, làm tròn 1 chữ số thập phân.
# Không in tên kho ở ô này.
```
